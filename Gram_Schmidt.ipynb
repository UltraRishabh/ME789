```python
import numpy as np
import time
```

```python
n = 1024
m = 1024
A = np.random.rand(m,n)
```

```python
R = np.zeros((n,n))
Q = np.zeros((m,n))
```

```python
start = time.time()

V = A.copy()

for i in range(n):
    R[i][i] = np.linalg.norm(V[:,i])
    # x += 2*m
    Q[:,i] = V[:,i]/R[i][i]
    # x+=m
    for j in range(i+1,n,1):
        R[i][j] = np.dot(Q[:,i],V[:,j])
        # x+= 2*m-1
        V[:,j] -= R[i][j]*Q[:,i]
        # x+=2*m
end = time.time()
```

```python
print("Approx: ",(2*m*n**2)/((end-start)*10**9),"\nExact: ",(2*m*n**2+3*m*n-n**2/2+n/2)/((end-start)*10**9))
```
