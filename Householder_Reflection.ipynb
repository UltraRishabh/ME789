```python
import numpy as np
import time
```

```python
m = 512
n = 512
A = np.random.rand(m,n)
I = np.identity(m)
```

```python
A
```

```python
k = min(m,n)
Q = np.identity(m)
vs = []     # Store Householder vectors instead of computing again for Q
start = time.time()
for i in range(k):
    x = A[i:m,i]
    sign = 1 if x[0] == 0 else np.sign(x[0])
    v = sign * np.linalg.norm(x) * I[i:m,i] + x             # 2(m-i)
    v = v / np.linalg.norm(v)                               # (m-i) + (m-i) + (m-i-1) + 1
    vs.append(v)
    A[i:m,i:n] -= 2 * np.outer(v, np.dot(v, A[i:m,i:n]))    # 4(m-i)*(n-i)

for i in range(k-1, -1, -1):          # i = k-1 downto 0 for Q backward sub
    v = vs[i]
    Q[i:m,i:m] -= 2 * np.outer(v, np.dot(v, Q[i:m,i:m]))    # 4(m-i)^2

end = time.time()
```
