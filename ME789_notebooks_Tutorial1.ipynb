```python
import random
import cProfile
import pstats
import time
import numpy as np
```

```python
def matrix_vector_multiply(A, x):

    rows = A.shape[0]

    y = np.zeros(rows)

    for i in range(rows):
        for j in range(rows):
            y[i] += A[i,j] * x[j]

    return y
```

```python
def main():

    # sizes = [256]
    sizes = [256, 512, 1024]
    print(f"{'Size':>6} {'Matrix Gen (s)':>15} {'Vector Gen (s)':>15} {'Multiply (s)':>15} {'Total (s)':>12}")
    #print(f"{'Size':>8} {'FLOPs':>15} {'Time(s)':>12} {'GFLOPS':>12}")
    print("-"*50)

    for N in sizes:
        # Matrix generation
        A = np.random.randint(0,100,(N,N))
        # Vector generation
        x = np.random.randint(0,100,N)
        start = time.perf_counter()
        result = matrix_vector_multiply(A, x)
        end = time.perf_counter()
        elapsed = end-start
        print(f"Time for matrix-vector multiplication ({N} x {N}) : {end - start:.6f} seconds")

        # Matrix-vector multiplication FLOPs
        flops = 2*N*N-N
        gflops = flops/(elapsed*1e9)
        print(f"{'Size':>8} {'FLOPs':>15} {'Time(s)':>12} {'GFLOPS':>12}")
        print(f"{N:8d} {flops:15.0f} {elapsed:12.6f} {gflops:12.6f}")


if __name__ == "__main__":
    main()
```

```python

def matrix_matrix_multiply(A, B):
    n = A.shape[0]

    C = np.zeros((n, n))

    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i, j] += A[i, k] * B[k, j]

    return C
```
