```python
import numpy as np
```

```python
pi = np.pi
r = 1
L = 10
# 0 <= h <= r
a = 0
b = r
TOL = 1e-5  # How small the interval [a,b] before termination
e = 0.00001  # Limit for func f(a) defined below
N = 1000
V = 12.4
```

```python
def f(h):
    return V - L*(0.5*pi*r*r - r*r*np.arcsin(h/r) - h*((r*r-h*h)**0.5))
```

```python
def bisection(a, b, TOL, e, N):
    p = 0
    FA = f(a)
    i = 0
    while(i<N):
        p = (a + b)/2
        FP = f(p)
        if abs(FP) < e or (b-a)/2 < TOL:
            print("value:",p)
            print("iters:",i+1)
            return p
        i+=1
        if FA*FP<0:
            b = p
        else:
            a = p
            FA = FP
    print("Max Iterations Reached, current p:",p)
    return
```

```python
h = bisection(a,b,TOL,e,N)
```

```python
print(f(h))
```
