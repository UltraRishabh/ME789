```python
import numpy as np
```

```python
## Diverges
def g1(x):
    return 0.5*((10/x - 4*x)**0.5)

def g2(x):
    return ((10/(4+x))**0.5)

def g3(x):
    return 0.5*((10 - x*x*x)**0.5)

def g4(x):
    return x - (x*x*x + 4*x*x - 10)/(3*x*x + 8*x)
```

```python
p = 1.5 # Initial Value
N = 1000 # Max Iters
e = 1e-9 # Tolerance

def FIP(p ,N ,g):
    i = 0
    x=p
    while(i<N):
        if abs(x - g(x)) < e:
            print("Value:",x)
            print("iters:",i+1)
            return x
        x = g(x)
        i+=1
    print("Value:",x,"\nMax Iters reached! (",N,")")
```

```python
FIP(p,N,g1)
FIP(p,N,g2)
FIP(p,N,g3)
FIP(p,N,g4)
```
